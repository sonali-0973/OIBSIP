## Missing Data Handling Strategy

The missing values are handled according to the meaning of each variable rather than applying one method to every column.

- **Categorical columns describing the absence of a feature** such as `Alley`, `BsmtQual`, `BsmtCond`, `BsmtExposure`, `BsmtFinType1`, `BsmtFinType2`, `FireplaceQu`, `GarageType`, `GarageFinish`, `GarageQual`, `GarageCond`, `PoolQC`, `Fence`, `MiscFeature`, and `MasVnrType` are filled with `"None"` because a missing value indicates that the house does not have that feature.

- **Numeric columns representing the size/capacity of an absent feature**, such as `MasVnrArea`, `BsmtFinSF1`, `BsmtFinSF2`, `BsmtUnfSF`, `TotalBsmtSF`, `BsmtFullBath`, `BsmtHalfBath`, `GarageCars`, and `GarageArea`, are filled with `0` when appropriate because no feature means zero area/capacity.

- `LotFrontage` is filled using the **median LotFrontage within each Neighborhood**. This preserves local housing characteristics and is more appropriate than using one global value.

- `GarageYrBlt` is filled with `0` when there is no garage. For remaining missing values, the median is used.

- `Electrical` has only a very small number of missing values, so the **mode** is used because it is a categorical variable.

This approach avoids unnecessary row deletion and preserves as much information as possible.

```python
cleaned_df = df.copy()

print("Created a copy for cleaning.")
```

```python
categorical_none_columns = [
    "Alley",
    "MasVnrType",
    "BsmtQual",
    "BsmtCond",
    "BsmtExposure",
    "BsmtFinType1",
    "BsmtFinType2",
    "FireplaceQu",
    "GarageType",
    "GarageFinish",
    "GarageQual",
    "GarageCond",
    "PoolQC",
    "Fence",
    "MiscFeature"
]

for column in categorical_none_columns:
    if column in cleaned_df.columns:
        cleaned_df[column] = cleaned_df[column].fillna("None")

print("Categorical missing values handled.")
```

```python
zero_fill_columns = [
    "MasVnrArea",
    "BsmtFinSF1",
    "BsmtFinSF2",
    "BsmtUnfSF",
    "TotalBsmtSF",
    "BsmtFullBath",
    "BsmtHalfBath",
    "GarageCars",
    "GarageArea"
]

for column in zero_fill_columns:
    if column in cleaned_df.columns:
        cleaned_df[column] = cleaned_df[column].fillna(0)

print("Feature-related numerical missing values filled with 0.")
```

```python
cleaned_df["LotFrontage"] = (
    cleaned_df.groupby("Neighborhood")["LotFrontage"]
    .transform(lambda x: x.fillna(x.median()))
)

# If any values are still missing, use the overall median
cleaned_df["LotFrontage"] = cleaned_df["LotFrontage"].fillna(
    cleaned_df["LotFrontage"].median()
)

print("LotFrontage missing values handled using neighborhood median.")
```

```python
# Houses without a garage are represented as 0
cleaned_df.loc[
    cleaned_df["GarageType"] == "None",
    "GarageYrBlt"
] = 0

# Any remaining missing values use the median
garage_year_median = cleaned_df.loc[
    cleaned_df["GarageYrBlt"] > 0,
    "GarageYrBlt"
].median()

cleaned_df["GarageYrBlt"] = cleaned_df["GarageYrBlt"].fillna(
    garage_year_median
)

print("GarageYrBlt missing values handled.")
```

```python
electrical_mode = cleaned_df["Electrical"].mode()[0]

cleaned_df["Electrical"] = cleaned_df["Electrical"].fillna(
    electrical_mode
)

print("Electrical missing values filled with mode:", electrical_mode)
```

```python
remaining_missing = cleaned_df.isnull().sum()

remaining_missing = remaining_missing[
    remaining_missing > 0
].sort_values(ascending=False)

print("Remaining missing values:")

if len(remaining_missing) == 0:
    print("No missing values remain.")
else:
    display(remaining_missing)
```

```python
duplicates_before = cleaned_df.duplicated().sum()

print("Duplicate rows found:", duplicates_before)

cleaned_df = cleaned_df.drop_duplicates()

duplicates_after = cleaned_df.duplicated().sum()

print("Duplicate rows after removal:", duplicates_after)
print("Rows after duplicate removal:", len(cleaned_df))
```

```python
# Identify categorical columns
categorical_columns = cleaned_df.select_dtypes(
    include=["object"]
).columns

# Remove leading/trailing whitespace
for column in categorical_columns:
    cleaned_df[column] = cleaned_df[column].str.strip()

# Convert empty strings to NaN
for column in categorical_columns:
    cleaned_df[column] = cleaned_df[column].replace("", np.nan)

# Check if standardization introduced any missing values
print("Missing values after standardization:")
print(cleaned_df.isnull().sum().sum())
```

## Data Standardisation

The categorical variables were checked for inconsistent formatting.

The dataset mainly uses standardized categorical codes such as `RL`, `RM`, `Pave`, `Gd`, `TA`, and `Normal`. Therefore, these values were not unnecessarily renamed because the codes have meaningful dataset-specific definitions.

Leading and trailing whitespace was removed from categorical columns, and empty strings were treated as missing values.

There are no actual date columns in the dataset requiring conversion to `datetime`. The variables `YearBuilt`, `YearRemodAdd`, `GarageYrBlt`, and `YrSold` represent years and are therefore retained as numeric year values.

```python
# ID should be treated as an identifier, not a numerical measurement
cleaned_df["Id"] = cleaned_df["Id"].astype(str)

# MSSubClass is a housing classification code rather than a continuous measurement
cleaned_df["MSSubClass"] = cleaned_df["MSSubClass"].astype(str)

# SalePrice is a monetary value
cleaned_df["SalePrice"] = cleaned_df["SalePrice"].astype(float)

print("Data types after correction:")
display(cleaned_df.dtypes.to_frame("Data Type"))
```

```python
print("Id data type:", cleaned_df["Id"].dtype)
print("MSSubClass data type:", cleaned_df["MSSubClass"].dtype)
print("SalePrice data type:", cleaned_df["SalePrice"].dtype)
```

```python
# Select numeric columns
numeric_columns = cleaned_df.select_dtypes(
    include=np.number
).columns

outlier_summary = []

for column in numeric_columns:

    Q1 = cleaned_df[column].quantile(0.25)
    Q3 = cleaned_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = cleaned_df[
        (cleaned_df[column] < lower_bound) |
        (cleaned_df[column] > upper_bound)
    ]

    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers)
    })

outlier_report = pd.DataFrame(outlier_summary)

outlier_report = outlier_report.sort_values(
    by="Outlier_Count",
    ascending=False
)

display(outlier_report)
```

```python
outliers_only = outlier_report[
    outlier_report["Outlier_Count"] > 0
]

display(outliers_only)
```

## Outlier Handling Decision

The IQR method was used to identify potential outliers in numeric variables.

An IQR outlier does not automatically mean that a value is incorrect. In a house-price dataset, unusually large houses, lots, garages, or sale prices can represent legitimate properties.

Therefore, the identified IQR outliers were reviewed rather than automatically deleted.

Values were retained when they were within a plausible real-world range for a house. Automatically removing all IQR outliers could remove legitimate high-value properties and introduce bias into the dataset.

No artificial values were created to reduce the number of outliers. The dataset retains legitimate extreme observations while documenting the detected outliers.

```python
important_columns = [
    "LotArea",
    "GrLivArea",
    "TotalBsmtSF",
    "1stFlrSF",
    "GarageArea",
    "SalePrice"
]

for column in important_columns:
    if column in cleaned_df.columns:
        print(f"\n{column}")
        print("-" * len(column))
        print("Minimum:", cleaned_df[column].min())
        print("Maximum:", cleaned_df[column].max())
```

```python
checks = {
    "Negative LotArea": (cleaned_df["LotArea"] < 0).sum(),
    "Negative SalePrice": (cleaned_df["SalePrice"] < 0).sum(),
    "Negative GrLivArea": (cleaned_df["GrLivArea"] < 0).sum(),
    "Invalid MoSold": (
        (~cleaned_df["MoSold"].between(1, 12))
    ).sum(),
    "Invalid YearSold": (
        (~cleaned_df["YrSold"].between(2000, 2020))
    ).sum()
}

for check, count in checks.items():
    print(f"{check}: {count}")
```
